# GRU 로 IMDB 리뷰 감성 분류하기
IMDB 영화 리뷰 데이터를 전처리한 뒤 

Embedding과 GRU 모델을 이용하여 리뷰의 감성을 긍정 또는 부정으로 분류

```python
# 정규표현식 모듈
import re\
# IMDB 영화 리뷰 데이터셋 불러오기
from tensorflow.keras.datasets import imdb
# 시퀀스 길이를 맞추기 위한 패딩 함수
from tensorflow.keras.preprocessing.sequence import pad_sequences
# 순차 모델 생성
from tensorflow.keras.models import Sequential
# 모델에 사용할 레이어
from tensorflow.keras.layers import Dense, GRU, Embedding
# 학습 조기 종료 및 모델 저장
from tensorflow.keras.callbacks import EarlyStopping, ModelCheckpoint
# 저장된 모델 불러오기
from tensorflow.keras.models import load_model

```

```python

# 리뷰 에서 자주 등장하는 상위 10,000개의 단어만 사용
vocab_size = 10000

# 리뷰 길이가 서로 다르기 때문에 최대 길이를 500으로 설정
max_len = 500

```

```python

# IMDB 데이터셋 불러오기 (학습 데이터 / 테스트 데이터)
(X_train, y_train), (X_test, y_test) = imdb.load_data(num_words=vocab_size)


# 리뷰 길이를 500으로 맞추기 위해 패딩 적용
X_train = pad_sequences(X_train, maxlen=max_len)
X_test = pad_sequences(X_test, maxlen=max_len)
```

```python
# 임베딩 벡터 차원 설정
embedding_dim = 100

# GRU 은닉층 유닛 수 설정
hidden_units = 128

# 순차 모델 생성
model = Sequential()

```

```python

# 단어를 벡터로 변환하는 임베딩 층
model.add(Embedding(vocab_size, embedding_dim))

# GRU 순환 신경망 층
model.add(GRU(hidden_units))

# 출력층 (이진 분류 → sigmoid 사용)
model.add(Dense(1, activation='sigmoid'))

```

```python

# 검증 손실이 증가하면 학습을 조기 종료
es = EarlyStopping(
    monitor='val_loss',
    mode='min',
    verbose=1,
    patience=4
)

# 검증 정확도가 가장 높은 모델 저장
mc = ModelCheckpoint(
    'GRU_model.h5',
    monitor='val_acc',
    mode='max',
    verbose=1,
    save_best_only=True
)

```

```python

# 모델 학습 설정
model.compile(
    optimizer='rmsprop',
    loss='binary_crossentropy',
    metrics=['acc']
)

# 모델 학습
history = model.fit(
    X_train,
    y_train,
    epochs=15,
    callbacks=[es, mc],
    batch_size=64,
    validation_split=0.2
)
```

```python
# 저장된 GRU 모델 불러오기
loaded_model = load_model('GRU_model.h5')

# 테스트 데이터 정확도 출력
print("\n 테스트 정확도 : %.4f" % (loaded_model.evaluate(X_test, y_test)[1]))

```

```python
from tensorflow.keras.datasets import imdb

word_to_index = imdb.get_word_index()
```

```python
def sentiment_predict(new_sentence):
    import re
    
    # 알파벳/숫자/공백만 남기고 소문자화
    new_sentence = re.sub('[^0-9a-zA-Z ]', '', new_sentence).lower()
    
    encoded = []
    
    # 단어 → 인덱스 변환
    for word in new_sentence.split():
        try:
            if word_to_index[word] <= 10000:
                encoded.append(word_to_index[word] + 3)
            else:
                encoded.append(2)  # <unk>
        except KeyError:
            encoded.append(2)  # 단어 사전에 없는 경우
    
    # 패딩
    pad_sequence = pad_sequences([encoded], maxlen=max_len)
    
    # 예측
    score = loaded_model.predict(pad_sequence)[0][0]
    
    if score > 0.5:
        print("{:.2f}% 확률로 긍정 리뷰입니다.".format(score * 100))
    else:
        print("{:.2f}% 확률로 부정 리뷰입니다.".format((1 - score) * 100))
```

```python
# 테스트할 영화 리뷰 문장 (부정적 리뷰)
test_input = ("This movie was just way too overrated. "
"The fighting was not professional and in slow motion. "
"I was expecting more from a 200 million budget movie."
" The little sister of T.Challa was just trying too hard to be funny. "
"The story was really dumb as well. "
"Don't watch this movie if you are going because others say its great unless "
"you are a Black Panther fan or Marvels fan.")

# 입력한 리뷰에 대해 감성 예측 수행
sentiment_predict(test_input)
```
