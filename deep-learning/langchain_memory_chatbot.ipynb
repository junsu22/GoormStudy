뉴스요약 서비스 구현

```python
# 시스템 프롬프트에 뉴스가 들어오면 주제, 요약, 가능 한 질문 세가지 출력 지정
def return_answer(input_text=''):
    system_prompt = """특정 문서가 입력되면 다음과 같은 형태로 문서를 분석하십시오.

1. 주어진 입력에 대해서 반드시 주제, 요약, 가능한 질문 이 세 가지를 순차적으로 작성해야 합니다.
2. 주제는 입력 문서의 주제를 한 줄로 요약합니다.
3. 요약은 입력 문서를 5줄로 요약합니다.
4. 가능한 질문은 입력 문서로부터 사람들이 할 수 있는 질문 세 가지를 파이썬 리스트 형태로 작성합니다.
5. 가능한 질문은 반드시 ["질문 1", "질문 2", "질문 3"]과 같이 파이썬 리스트 형태로 작성되어야 하는 점에 유의하십시오.

이제 시작합니다.
"""

    user_content = "입력: " + input_text + "\n주제:"

    response = client.chat.completions.create(
        model="gpt-4o",
        messages=[
            {"role": "system", "content": system_prompt},
            {"role": "user", "content": user_content}
        ]
    )

    return response.choices[0].message.content

# 뉴스기사를 입력에 넣음
test = '''새처럼 움직이는 항공 로봇 '레이븐'.

스위스 로잔 연방 공대(EPFL) 신원동 박사 팀이 개발했는데요.

연구팀은 새의 엉덩이와 발목, 발을 참고해 뒷다리를 만들었습니다.

실제 새처럼 땅에서 걷고, 틈새를 건너뛰고, 위로 뛰어오릅니다.

껑충 뛰면서 날아오르는 이 기능이 핵심인데요.

기존 고정익 항공기처럼 지상을 달리지 않고 제자리에서 날아오릅니다.

연구팀은 레이븐의 점프 이륙이 점프하지 않고 날아오를 때보다 에너지 효율이 더 높다고 설명했는데요.

또 기존 고정익 항공기에 이 기능을 도입하면 복잡한 지형에서도 효율적으로 쓸 수 있을 거라고 덧붙였습니다.

#새 #로봇 #항공 #공대 #스위스 #신원동 #박사 #점프
'''

result = return_answer(input_text=test)
print(result)
```

```python
# 시스템 프롬프트 중간에 [Example] 이라는 내용과 함께 사용자의 입력과 답변을 예시로 기재
def return_answer(input_text):
    system_prompt = """You are an expert at extracting keywords from a given sentence. Extract keywords from a given sentence that are key to the context.

[Example]
input: '외신에 따르면 중국 이카이 글로벌 보도를 인용해 모더나가 중국 전용 mRNA 백신 개발을 위해 중국에 최대 10억 달러(약 1조 3017억 원)를 투자하기로 했다고 보도했다.'
keyword: ['모더나', 'mRNA', '1조 3017억 원']

Now the sentence you want to extract keywords from. The output format must be a list in Python.
"""

    user_content = "input: " + input_text + "\nkeyword: "

    response = client.chat.completions.create(
        model="gpt-4o",
        messages=[
            {"role": "system", "content": system_prompt},
            {"role": "user", "content": user_content}
        ],
    )

    return response.choices[0].message.content

# 입력
text = '''3일 업계에 따르면 서울 신라호텔의 더 파크뷰는 12월 120일 저녁 가격을 19만 5000원으로, 2131일 저녁 가격을 21만 5000원으로 각각 인상한다. 평소 평일·주말 저녁 가격이 18만 5000원이던 것을 고려하면 각각 1만 원과 3만 원 올랐다. 가격 인상에도 불구하고 더 파크뷰의 12월 주말 예약은 대부분 찼고 12월 평일 예약 역시 빠르게 마감되고 있다고 신라호텔 측은 전했다.

롯데호텔 서울의 뷔페 라세느는 12월 평일·주말 저녁 가격을 19만 원으로 기존 대비 1만 원 올린다. 크리스마스 연휴 때인 2325일과 연말 3031일 저녁 가격은 20만 5000원이다. 롯데호텔 서울의 라세느도 현재 크리스마스 이브인 12월 24일과 크리스마스 당일인 12월 25일 예약은 대부분 마감됐다.

워커힐 호텔앤리조트는 그랜드 워커힐에서 운영하던 더뷔페를 비스타 워커힐로 확장 이전해 이날 가격을 올렸다. 평일·주말 저녁 가격은 15만 9000원에서 18만 9000원으로 3만 원 인상됐다. 다만 워커힐은 이번 가격 책정에 인플레이션(물가 상승)을 미리 반영한 만큼 연말을 앞두고 추가 인상을 단행하지 않을 계획이라고 밝혔다.

그랜드 인터컨티넨탈 서울 파르나스와 인터컨티넨탈 서울 코엑스도 다음주께 인상된 가격을 공지할 계획이다.
'''

result = return_answer(input_text=text)
print(result)
```

감성 분석 서비스


주어진 뉴스 기사가 긍정적인 뉴스인지 부정적인 뉴스인지

둘 다 아닌 중립에 해당하는 뉴스 기 사인지를 판단하는 감성 분석 서비스

```python
# 가상의 회사 씨씨에스라는 회사가 주가가 상한가라는 텍스트를 함수의 입력으로 사용하여 답변
def return_label(text):
    prompt = '''주어진 텍스트가 긍정인지 중립인지 부정인지 예측하시오. 당신의 답변은 오로지 [positive, negative, neutral] 셋 중 하나여야만 합니다.

ex) 'SK하이닉스가 2분기 실적에서 역대 최고를 기록하였다' => "positive"
'''

    response = client.chat.completions.create(
        model="gpt-4o",
        messages=[
            {"role": "system", "content": prompt},
            {"role": "user", "content": text + " => "}
        ]
    )

    return response.choices[0].message.content


result = return_label(
    '씨씨에스로고 씨씨에스 주가가 상한가다. 3일 기준 씨씨에스는 29.84%(825원) 상승한 3590원에 거래를 마쳤다. 상온 초전도체 이슈가 부각되면서 주가가 연일 영향을 받는 모양새다.'
)

print(result)
```

```python
# 코사인 유사도 구하기 -1~ 1
# 벡터 A, B 의 코사인
import numpy as np
from numpy import dot
from numpy.linalg import norm

def cos_sim(A, B):
    return dot(A, B) / (norm(A) * norm(B))

vec1 = np.array([0, 1, 1, 1])
vec2 = np.array([1, 0, 1, 1])
vec3 = np.array([2, 0, 2, 2])

print('벡터 1과 벡터 2의 유사도:', cos_sim(vec1, vec2))
print('벡터 1과 벡터 3의 유사도:', cos_sim(vec1, vec3))
print('벡터 2와 벡터 3의 유사도:', cos_sim(vec2, vec3))
```

```python
# 임베딩 유사도 테스트를 위한 예제 문장 데이터 생성
# 다양한 의미(배: 음식 / 배: 선박 / 스팀: 게임 / 스팀: 요리)를 포함하도록 구성

data = [
    "저는 배가 고파요",
    "저기 배가 지나가네요",
    "굶어서 허기가 지네요",
    "허기워기라는 게임이 있는데 즐거워",
    "스팀에서 재밌는 거 해야지",
    "스팀 에어프라이어로 연어구이 해 먹을 거야"
]

# 데이터프레임 생성 (text 컬럼에 문장 저장)
df = pd.DataFrame(data, columns=["text"])

df
```

```python
# 각 문장을 임베딩 모델을 통해 벡터로 변환하는 함수 정의
def get_embedding(text):
    return embeddings.embed_query(text)

# DataFrame의 text 컬럼에 대해 임베딩 수행
# 각 문장을 벡터로 변환하여 새로운 'embedding' 컬럼에 저장
df["embedding"] = df.apply(
    lambda row: get_embedding(row.text),
    axis=1
)

df
```

```python
# 코사인 유사도 계산 함수
# 두 벡터(A, B) 간의 유사도를 계산 (값이 1에 가까울수록 유사)
def cos_sim(A, B):
    return dot(A, B) / (norm(A) * norm(B))


# 질의(query)와 가장 유사한 문서를 찾는 함수
def return_answer_candidate(df, query):

    # 입력된 query를 임베딩 벡터로 변환
    query_embedding = get_embedding(query)

    # 각 문서의 임베딩과 query 임베딩 간의 유사도 계산
    # 계산된 유사도를 'similarity' 컬럼에 저장
    df["similarity"] = df["embedding"].apply(
        lambda x: cos_sim(np.array(x), np.array(query_embedding))
    )

    # 유사도가 높은 순으로 정렬 후 상위 3개 문서 선택
    top_three_doc = df.sort_values("similarity", ascending=False).head(3)

    return top_three_doc
```

```python
# 질의 문장을 입력하여 유사한 문장 검색
sim_result = return_answer_candidate(
    df,
    "아무것도 안 먹었더니 꼬르륵 소리가 나네"
)

sim_result
```

# 백분위수 방식
유사도를 비교 하고 , 진짜 크게 바뀌는 지점을 자른다.
ex. 상위 5%

“얼마나 민감하게 자를지 결정하는 기준”


95 = 크게 바뀔 때만 자름


50 = 조금만 달라도 자름

```python
# 의미 기준으로 텍스트를 나누는 청커 생성
# percentile=95 → 의미가 크게 바뀌는 상위 5% 지점에서만 자름 (덜 자름, 덩어리 큼)

text_splitter = SemanticChunker(
    OpenAIEmbeddings(),
    breakpoint_threshold_type="percentile",
    breakpoint_threshold_amount=95,
)

# 텍스트를 의미 단위로 분할
texts = text_splitter.create_documents([file])

# 분할된 청크 개수 출력
print("분할된 청크의 수:", len(texts))
```

# 표준편차 방식
평균에서 얼마나 벗어났는지 , 벗어난 기준으로 자른다.


예시 ) 점수: 80, 82, 79, 81, 150 ---> 150 너무 멀리 떨어짐

청킹 적용시 예시

A: 오늘 날씨 좋다


B: 산책 나갔다


C: 기분 좋다


D: 주식 폭락했다  ← 갑자기 다른 주제

```python
# 의미 기준으로 텍스트를 나누는 청커 생성
# standard_deviation=3 → 평균보다 3배 이상 크게 의미가 바뀌는 지점에서만 자름 (아주 덜 자름)

text_splitter = SemanticChunker(
    OpenAIEmbeddings(),
    breakpoint_threshold_type="standard_deviation",
    breakpoint_threshold_amount=3,
)

# 텍스트를 의미 단위로 분할
texts = text_splitter.create_documents([file])

# 분할된 청크 개수 출력
# standard_deviation=3 → 평균보다 3배 이상 차이날 때만 자름
print("분할된 청크의 수:", len(texts))
```

# 사분위수 방식

갑자기 튀는 값 = 이상치


하위 25% | 중간 50% | 상위 25%


breakpoint_threshold_amount=1.5

일반 범위(중간 값들)에서 벗어나면 자름

```python
# 의미 기준으로 텍스트를 나누는 청커 생성

text_splitter = SemanticChunker(
    OpenAIEmbeddings(),
    breakpoint_threshold_type="interquartile",
# interquartile=1.5 → 일반 범위를 벗어나는 지점(이상치)에서 자름
    breakpoint_threshold_amount=1.5,
)

# 텍스트를 의미 단위로 분할
texts = text_splitter.create_documents([file])

# 분할된 청크 개수 출력
print("분할된 청크의 수:", len(texts))
```

```python
import os  # 환경변수 설정 (API 키 등)

from langchain_openai import ChatOpenAI  # OpenAI GPT 모델 사용
from langchain_core.prompts import PromptTemplate  # 프롬프트 템플릿
from langchain_core.runnables.history import RunnableWithMessageHistory  # 대화 기록 유지
from langchain_community.chat_message_histories import ChatMessageHistory  # 메시지 저장 객체
from langchain_core.tracers.stdout import ConsoleCallbackHandler  # 실행 로그 출력
```

```python
# OpenAI API 키 설정
os.environ["OPENAI_API_KEY"] = ""

# LLM(모델) 객체 생성
llm = ChatOpenAI(
    model_name="gpt-4o",   # 사용할 모델 이름
    temperature=0.1,       # 창의성 (낮을수록 정확, 높을수록 창의적)
    max_tokens=2048        # 최대 응답 길이
)
```

```python
# 질의 내용 정의
question = "세종대왕이 누구인지 설명해주세요"

# 모델에 질문하고 결과 받기
result = llm.invoke(question)

# 결과 출력
print(result.content)
```

```python
# 질문 템플릿 정의 (변수 포함)
template = "{who}가 누구인지 설명해주세요"

# 템플릿 객체 생성
prompt = PromptTemplate(
    template=template,
    input_variables=["who"]  # 사용할 변수
)

# 템플릿 구조 확인 (미리보기)
print(prompt)
```

```python
# 템플릿에 값(who)을 넣어서 실제 문장 생성
print(prompt.format(who="오바마"))
```

```python
# 프롬프트와 LLM을 연결하여 체인 생성
llm_chain = prompt | llm

# 체인 실행 (변수 값 전달)
# invoke : 실행함수
result = llm_chain.invoke({"who": "이순신 장군"})

# 결과 출력
print(result.content)
```

```python
# 체인 실행 + 실행 과정 로그 출력
result = llm_chain.invoke(
    {"who": "이순신 장군"},
    config={"callbacks": [ConsoleCallbackHandler()]}
)
```

```python
# 프롬프트 템플릿 생성 (대화형 챗봇용)
# 이전 대화 + 현재 질문 → 프롬프트 생성 → GPT → 답변
# 대화 문맥 포함해서 답변하게 만드는 프롬프트

template = """아래는 사람과 AI의 친근한 대화입니다.
AI의 이름은 위키독스봇입니다.
이 AI는 위키독스에서 만들어졌습니다.
대화 문맥을 바탕으로 친절하게 답변하세요.

Current Conversation: {history}
Human: {input}
AI:"""

prompt = PromptTemplate(
    template=template,
    input_variables=["history", "input"]  # 이전 대화 + 현재 질문
)

# LLM 객체 생성
llm = ChatOpenAI(model_name="gpt-4o")

# 프롬프트와 LLM 연결 (체인 생성)
chain = prompt | llm
```

```python
store = {}
session_id = "test"
```

```python
# 아주 중요 : 사람 대화 섞일 수 있음 . 그러니 숙지할 것

# 사용자 세션이 없으면 새로 만들고 있으면 , 있으면 기존 대화 가져오기

# 해당 session_id가 없으면 새로운 대화 기록 생성
if session_id not in store:
    store[session_id] = ChatMessageHistory()

# 현재 세션의 대화 기록 가져오기
session_history = store[session_id]
```

```python
# 체인에 대화 기록(메모리)을 연결
# 이전 대화(history) + 현재 질문(input)을 같이 넣어서 답변하도록 설정
with_message_history = RunnableWithMessageHistory(
    chain,
    lambda session_id: session_history,  # session_id에 해당하는 대화 기록 반환
    input_messages_key="input",          # 현재 사용자 입력 key
    history_messages_key="history"       # 이전 대화 기록 key
)
```

```python
# 메모리를 포함한 체인 실행 (대화 유지됨)
result = with_message_history.invoke(
    {"input": "당신은 어디에서 만들어졌습니까?"},
    config={"configurable": {"session_id": "test"}}  # 사용자 구분 (세션)
)

# 결과 출력
print(result.content)
```


[대화 기록 저장 구조 설명]

- store 딕셔너리에 session_id를 key로 사용하여 대화 기록을 저장한다.
- 현재 예시에서는 "test"라는 session_id에 대한 대화가 저장되어 있다.

■ 구조
store = {
    "test": InMemoryChatMessageHistory(
        messages=[
            HumanMessage(content="사용자 질문"),
            AIMessage(content="AI 답변"),
            ...
        ]
    )
}

■ 구성 요소
- HumanMessage: 사용자의 입력 (질문)
- AIMessage: AI의 응답 (답변)

■ 동작 방식
1. 사용자가 질문을 입력하면 HumanMessage로 저장된다.
2. AI가 답변을 생성하면 AIMessage로 저장된다.
3. 이 메시지들이 순서대로 쌓이며 대화 기록(history)이 된다.
4. 이후 질문 시 이전 대화(history)가 함께 전달되어 문맥을 유지한다.

■ 핵심
- session_id 기준으로 사용자별 대화가 분리된다.
- 이 구조 덕분에 "이전 대화 기억하는 챗봇" 구현이 가능하다.
