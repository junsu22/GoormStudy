```
# 라이브러리 호출
import torch
import torch.nn as nn
# GPU 연산이 수행가능하다면 GPU 연산 수행
device = 'cuda' if torch.cuda.is_available() else 'cpu'
torch.manual_seed(777)
if device == 'cuda':
    torch.cuda.manual_seed_all(777)

```

```
# 다층 퍼셉트론으로 XOR 문제 풀기
# 다층 퍼셉트론 설계(MLP)
# 입력층, 은닉층 1, 2, 3, 출력층
X = torch.FloatTensor([[0, 0], [0, 1], [1, 0], [1, 1]]).to(device)
Y = torch.FloatTensor([[0], [1], [1], [0]]).to(device)
```

```
model = nn.Sequential(
    nn.Linear(2, 10),
    nn.Sigmoid(),
    nn.Linear(10, 10),
    nn.Sigmoid(),
    nn.Linear(10, 10),
    nn.Sigmoid(),
    nn.Linear(10, 1),
    nn.Sigmoid()
)
```

```
# XOR 문제에 사용할 입력값(X)과 정답값(Y)을 텐서로 만들기
X = torch.FloatTensor([[0, 0], [0, 1], [1, 0], [1, 1]]).to(device)
Y = torch.FloatTensor([[0], [1], [1], [0]]).to(device)
```

```
model = nn.Sequential(
    nn.Linear(2, 10),
    nn.Sigmoid(),
    nn.Linear(10, 10),
    nn.Sigmoid(),
    nn.Linear(10, 10),
    nn.Sigmoid(),
    nn.Linear(10, 1),
    nn.Sigmoid()
)
```

```
# 비용함수, 옵티마이저 선언
criterion = torch.nn.BCELoss().to(device)
optimizer = torch.optim.SGD(model.parameters(), lr=1)
```

```
# 학습
# 총 10,001번의 에포크를 수행하며, 각 에포크마다 역전파를 수행
for step in range(10001):
    optimizer.zero_grad()

    # forward 연산
    hypothesis = model(X)

    # 비용함수 계산
    cost = criterion(hypothesis, Y)
    cost.backward()

    optimizer.step()

    # 100의 배수에 해당하는 에포크마다 비용 출력
    if step % 100 == 0:
        print(step, cost.item())
```

```
# 학습된 다층 퍼셉트론의 값 예측
# 모델이 정말 XOR 문제를 풀 수 있는지 테스트
with torch.no_grad():
    hypothesis = model(X)
    predicted = (hypothesis > 0.5).float()
    accuracy = (predicted == Y).float().mean()

    print('모델의 출력값(Hypothesis): ', hypothesis.detach().cpu().numpy())
    print('모델의 예측값(Predicted): ', predicted.detach().cpu().numpy())
    print('실제값(Y): ', Y.cpu().numpy())
    print('정확도(Accuracy): ', accuracy.item())
```
