```
# Stack
# 자료를 쌓아놓은 형태, 책쌓기, 후입선출(LIFO)

# Stack 클래스 만들기
class Node:
    def __init__(self, data):
        self.data = data  # 데이터 저장
        self.next = None  # 다음 노드 가리키기

class Stack:
    def __init__(self):
        self.top = None  # 스택의 맨 위 (head와 비슷한 역할)

    def push(self, data):
        """스택에 데이터 추가 (맨 위에 쌓기)"""
        if self.top is None:  # 스택이 비어있으면
            self.top = Node(data)
        else:  # 스택에 이미 데이터가 있으면
            node = Node(data)
            node.next = self.top  # 새 노드가 기존 top을 가리킴
            self.top = node  # top을 새 노드로 변경

    def pop(self):
        """스택에서 데이터 제거 (맨 위에서 꺼내기)"""
        if self.top is None:  # 스택이 비어있으면
            return None
        node = self.top  # 현재 top 저장
        self.top = self.top.next  # top을 다음 노드로 이동
        return node.data  # 꺼낸 데이터 반환

    def peek(self):
        """스택의 맨 위 데이터 확인 (제거하지 않음)"""
        if self.top is None:
            return None
        return self.top.data

    def is_empty(self):
        """스택이 비어있는지 확인"""
        return self.top is None


# 테스트 코드
if __name__ == "__main__":
    s = Stack()

    # A, B, C 순서로 push
    for i in range(3):
        s.push(chr(ord("A") + i))
        print(f"Push data = {s.peek()}")
    print()

    # 스택이 빌 때까지 pop (C, B, A 순서)
    while not s.is_empty():
        print(f"Pop data = {s.pop()}")
    print()

    # 빈 스택에서 peek (None 출력)
    print(f"Peek data = {s.peek()}")
```

```
# stack 을 이용하여 문자 거꾸로 출력하기
def reverse_word(word : str) -> str:
  answer: str = ""
  s = Stack()
  for w in word:
    s.push(w)
  while not s.is_empty():
    answer += s.pop()
  return answer

# 테스트 코드
print(reverse_word("aircraft"))
```

```
# 빠진 괄호 (짝이 맞는지 검사)
def check_bracket(
    expression: str,
) -> bool:  # expression의 각 문자를 하나씩 본다 / 타입은 True/F
    s = Stack()

    for exp in expression:
        if exp == "(":  # 보고있는 괄호가 여는 괄호라면
            s.push(exp)  # 스택(s)에 담는다(push)

        # elif exp == ")":
        #     if not s.pop():
        #         return False

#=====트러블 슈팅========
# ''' (docstring 용도) : 기존 의 코드와 비교해보려 사용.
# 주석과 비슷한 느낌인 줄 알고 편리하게 주석처럼 사용하다가 ,
# 들여쓰기에 영향을 주어 에러를 발생시켰다.
# =======================


        # 개선한 코드
        elif exp == ")":
            if s.is_empty():  # 스택이 비었는지 확인
                return False
            s.pop()

    return True if s.is_empty() else False


# 테스트 코드
print(check_bracket("((a * (b + c)) -d) / e"))
print(check_bracket("(((a* (b + c)) -d) / e"))
```

```
# 응용버전

def check_bracket2(expression : str) -> bool:
    brackets : dict[str, str] = {")": "(", "}":"{", "]":"["}
    s = Stack()

    for exp in expression:
        if exp in brackets.values():  # 여는 괄호
            s.push(exp)

        elif exp in brackets:  # 닫는 괄호
            if s.is_empty():  #  안정성 . 비어있는지 먼저 확인하는 코드 추가
                return False

            pop_bracket = s.pop()  # pop
            if pop_bracket != brackets[exp]:  # 짝 확인
                return False

    return s.is_empty()

# 테스트 코드
print(check_bracket2("[{a * (b + c)} - d ]/ e"))
print(check_bracket2("[{a * (b + c)] - d ] / e"))
```
