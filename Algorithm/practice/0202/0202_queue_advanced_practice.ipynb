```
# Queue class 구현
class Node:
  def __init__(self, data):
    self.data = data  # 데이터 저장
    self.next = None  # 다음 노드 가리키기

class Queue:
  def __init__(self):
    self.front = None  # 큐의 맨 앞 (head 역할)
    self.rear = None   # 큐의 맨 뒤

  def enqueue(self, data):  # 큐에 데이터 추가 (맨 뒤에 넣기)
    if self.front is None:  # 큐가 비어있으면
      self.front = self.rear = Node(data)  # front와 rear가 같은 노드
    else:  # 큐에 데이터가 있으면
      node = Node(data)
      self.rear.next = node  # 기존 rear 다음에 새 노드 연결
      self.rear = node  # rear를 새 노드로 이동

  def dequeue(self):  # 큐에서 데이터 제거 (맨 앞에서 빼기)
    if self.front is None:  # 큐가 비어있으면
      return None
    node = self.front  # 현재 front 저장
    if self.front == self.rear:  # 노드가 1개만 있으면
      self.front = self.rear = None  # 큐 비우기
    else:  # 노드가 여러 개 있으면
      self.front = self.front.next  # front를 다음으로 이동
    return node.data  # 제거한 데이터 반환

  def is_empty(self):  # 큐가 비어있는지 확인
    return self.front is None

# # 테스트 코드
# if __name__ == "__main__":
#   q = Queue()

#   for i in range(3):  # A, B, C 순서로 enqueue
#     q.enqueue(chr(ord("A") + i))
#     print(f"Enqueue data = {q.rear.data}")
#   print()

#   while not q.is_empty():  # 큐가 빌 때까지 dequeue (A, B, C 순서)
#     print(f"Dequeue Data = {q.dequeue()}")
```

```
'''
요새푸스 순열
1번부터 N 번까지 N명의 사람이 원을 이루면서 앉아있고,
양의 정수 K(≤ N) 가 주어진다.
이제 순서대로 K 번째 사람을 제거한다.
한 사람이 제거되면 남은 사람들로 이루어진 원을 따라 이 과정을 계속해 나간다.
이 과정은 N 명의 사람이 모두 제거될 때 까지 계속 된다. 원에서 사람들이 제거 되는 순서를 (N,K)-
요세푸스 순열 이라고 한다. 예를 들어 (7,3)- 요세푸스 순열은 <3,6,2,7,5,1,4>이다.

N과 K가 주어지면 (N,K)-요세푸스 순열을 구하는 프로그램을 작성하시오.

ex.
입력 : 7,3
출력 : [3,6,2,7,5,1,4]

'''

# 리스트를 이용한  풀이 . 직관적이나 , 인덱스를 계속 순회해야함.
def josephus(n: int, k: int) -> list[int]:
  res: list[int] = []
  line : list[int] = [1 for _ in range(n)] # 값이 1 이면 제거대상. 0이면 제거 완료
  i : int = -1
  for _ in range(n-1):
    count: int = 0
    while count < k:
      i = (i + 1) % n # 계속 순환하므로 나머지 연산으로 인덱스 계산
      if line[i]: # 제거된 것이 아닐 때만 count 증가
        count += 1
    line[i] = 0
    res.append(i+1)

  res.append(line.index(1)+1)
  return res

# 테스트 코드
print (josephus(10, 7))




```

```
#  큐를 이용하여 요세푸스 풀기
# 원형 큐 자연스럽게 표현이 되나 코드가 리스트 보다 길어질 수 있다.

def josephus_q(n: int, k: int) -> list[int]:
    res: list[int] = [] # 제거 순서를 저장
    q = Queue()  # 직접 구현한 큐 클래스 사용

    for i in range(n):  # 1부터 n까지 큐에 추가
        q.enqueue(i + 1)

    while q.front.next:  # 끝에서 두 번째 원소까지만 반복
        for _ in range(k - 1):  # k-1번 회전  (맨 앞을 맨 뒤로)
            q.enqueue(q.dequeue())
        res.append(q.dequeue())  # k번째 원소 제거

    res.append(q.dequeue())  # 마지막 남은 원소
    return res


# 테스트 코드
print(josephus_q(10, 7))
```

```
# 큐 뒤집기
# 스택 이용방법
def reverse_queue(q: Queue) -> None:
  s = []  # 스택으로 순서 뒤집기
  while not q.is_empty(): # 큐 -> 스택
    s.append(q.dequeue())
  while s:  # 스택 -> 큐(순서 반대로)
    q.enqueue(s.pop())

# 큐 출력 함수 (상태 출력 메서드가 없어, 별도 함수를 만들어 출력)
def display(q: Queue) -> None:
  node = q.front
  while node:
    print(node.data, end = " ")
    node = node.next
  print()

# 테스트 코드
# 원본
q = Queue()
for i in range(5):
  q.enqueue(i)
display(q)
# 뒤집기
reverse_queue(q)
display(q)
```

```
# 재귀를 이용한 큐 뒤집기
# 큐가 빌 때까지 반복
# 큐의 자료를 하나 꺼내서 변수에 보관
# 자신을 호출하여 나머지 자료도 꺼냄
# 재귀에서 돌아오면서 큐에 자료를 넣음 (역순)


def reverse_queue_recursive(q:Queue) -> None:
  if not q.is_empty():
    data = q.dequeue()
    reverse_queue_recursive(q)
    q.enqueue(data)


# 테스트 코드
# 원본
q = Queue()
for i in range(5):
  q.enqueue(i)
display(q)

# 뒤집기
reverse_queue_recursive(q)
display(q)
```
