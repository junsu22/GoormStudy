```
# Queue class 구현
class Node:
  def __init__(self, data):
    self.data = data  # 데이터 저장
    self.next = None  # 다음 노드 가리키기

class Queue:
  def __init__(self):
    self.front = None  # 큐의 맨 앞 (head 역할)
    self.rear = None   # 큐의 맨 뒤

  def enqueue(self, data):  # 큐에 데이터 추가 (맨 뒤에 넣기)
    if self.front is None:  # 큐가 비어있으면
      self.front = self.rear = Node(data)  # front와 rear가 같은 노드
    else:  # 큐에 데이터가 있으면
      node = Node(data)
      self.rear.next = node  # 기존 rear 다음에 새 노드 연결
      self.rear = node  # rear를 새 노드로 이동

  def dequeue(self):  # 큐에서 데이터 제거 (맨 앞에서 빼기)
    if self.front is None:  # 큐가 비어있으면
      return None
    node = self.front  # 현재 front 저장
    if self.front == self.rear:  # 노드가 1개만 있으면
      self.front = self.rear = None  # 큐 비우기
    else:  # 노드가 여러 개 있으면
      self.front = self.front.next  # front를 다음으로 이동
    return node.data  # 제거한 데이터 반환

  def is_empty(self):  # 큐가 비어있는지 확인
    return self.front is None

# 테스트 코드
if __name__ == "__main__":
  q = Queue()

  for i in range(3):  # A, B, C 순서로 enqueue
    q.enqueue(chr(ord("A") + i))
    print(f"Enqueue data = {q.rear.data}")
  print()

  while not q.is_empty():  # 큐가 빌 때까지 dequeue (A, B, C 순서)
    print(f"Dequeue Data = {q.dequeue()}")
```

```
# 큐를 이용한 이진수 만들기
def generate_binary2(n: int) -> None:
    q = Queue()  # 큐 생성
    q.enqueue("1")  # 시작값 "1"

    for _ in range(n):
        i = q.dequeue()  # 큐에서 이진수 꺼내기
        q.enqueue(i + "0")  # 뒤에 "0" 붙여서 추가
        q.enqueue(i + "1")  # 뒤에 "1" 붙여서 추가
        print(i)

# 테스트 코드
generate_binary2(10)
```
