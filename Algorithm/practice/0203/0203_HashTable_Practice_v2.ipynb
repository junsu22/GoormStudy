```
"""
해시테이블에 대한 예제를 구현하고 과정을 설명하세요
해시테이블의 충돌처리기법에 대해 설명하고 구현하세요
트리와 이진트리의 구성에 대해 이해하세요
트리 순회 방법에 따라 노드 추가/삭제 기법에 대해 예제를 구현하고 과정을 설명하세요

"""


# 해시테이블 클래스 만들기
# 게임으로 생각해보기
# <해시테이블이란 새로운 직업을 만들기>
class HashTable:
    def __init__(self, length=5):  # 캐릭터 생성할때 하는 기본세팅(자신, 인벤5칸)
        self.max_len = length  # 인벤토리 크기 정하기
        self.table = [
            [] for _ in range(self.max_len)
        ]  # 빈 칸들 만들기. 처음시작하면 아무것도 없으니까.

    # 설명 : 새로운 직업을 만들기 위한 기본셋팅, 빈인벤토리 5칸을 만들었다

    # 크기 줄여주기
    def _hash(self, key):  # "_"는 내부함수를 뜻함.
        res = sum([ord(s) for s in key])
        # 1. ord : 문자를 숫자로 바꿔주는 함수
        # 2. sum : 숫자를 전부 다 더한다.

        # 인벤토리 칸 개수로 나눈 나머지 = 저장할 칸 번호
        # 나누는 이유 : 숫자를 다 합산해서 엄청 큰 숫자가 나오지만
        # 만든 칸은 5개. 큰 숫자 (ex.100 번째 칸이 없어서.
        # 범위 안에 들어 갈 수 있게 나눠주는 거임 (0~4 안에 들어 갈수 있도록)
        # 다 더한값 % 만든 인벤 의 개수
        return res % self.max_len

    # 설명 : 템들이 모두 칸에 들어 갈 수 있도록 나눠서 크기를 줄여줬다.

    # 세팅
    def set(self, key, value):
        # 어느 칸에 넣을지 게산
        index = self._hash(key)
        # 그 칸에 템과 수량을 추가한다.
        self.table[index].append((key, value))

    # 템과 수량을 계산해서 몇번째 슬롯에 넣을지 정했다.

    # 지금 까지 한 일 .
    # 캐릭터 생성해서 인벤만들어주고 물약 계산해서 슬롯에 넣어줬음.

    def get(self, key):
        index = self._hash(key)  # 몇번째 칸에 있을지 계산
        value = self.table[index]  # 그 칸 전체를 가져온다

        if not value:  # 값이 비었는지
            return None  # 없다면 None 리턴

        for v in value:  # 그 칸의 아이템들 하나씩 확인
            if v[0] == key:  # 이름이 같은지 확인 (충돌 시 여러 개 중에서 찾기)
                return v[1]  # 수량 리턴

        return None  # 다 찾아봤지만 없음

    # 지금까지 한 일:
    # 캐릭터 만들고, 인벤 준비하고, 아이템 넣고(set), 아이템 찾기(get) 완성!
    # 충돌 처리: 같은 칸에 여러 개 들어가면 리스트로 저장하고, 찾을 때 하나씩 확인


# 앞에서 구현한 클래스 사용
if __name__ == "__main__":
    capital = HashTable()  # 객체 생성
    country = [
        "Korea",
        "America",
        "China",
        "England",
        "turkiye",
    ]  # 국가이름 5개 , 도시 이름 5개 매칭
    city = ["Seoul", "Washington", "Beijing", "London", "Ankara"]

    for co, ci in zip(country, city):
        capital.set(co, ci)
        # co , ci = ("korea" , "seoul")  해시 테이블에 저장
        #     ("merica", "Washington") ... 루프가 돌면서 해시 테이블 에 저장
    print("해시 테이블의 상태")
    print("=================")
    for i, v in enumerate(capital.table):
        # i : 인덱스(0 ,1, 2, 3, 4)
        # v : 값
        print(i, v)
    print()

    print("======================")
    # 문자를 숫자로 바꾼 전체 합을 다 더해서 5로 나누었을때 나머지는 0
    # 'A'+'m'+'e'...  65 + 109 + 101 .. %5 = 0 America = 0번째
    print(f"Capital of America = {capital.get('America')}")
    print(
        f"Capital of Korea = {capital.get('Korea')}"
    )  # 498% 5 = 3 나머지 3 (중국 과 충돌)
    print(f"Capital of England = {capital.get('England')}")
    print(f"Capital of China = {capital.get('China')}")  # 483 % 5 = 3 나머지 3
    # 위에 country , city 에서 japan은 선언 하지 않아 결과는 None
    print(f"Capital of Japan = {capital.get('Japan')}")
    print(f"Capital of turkiye = {capital.get('turkiye')}")
```

```
# 해시테이블 클래스 만들기 - 다른 방법
# 내장함수 hash 이용하기 (파이썬에 내장된 함수)

class HashTable:
    def __init__(self, length=5):
        self.max_len = length
        self.table = [[] for _ in range(self.max_len)]

    # 직접 만든 _hash 는 쓰지 않음. 내장 된 함수를 이용할 거임.
    # def _hash(self, key):
    #     res = sum([ord(s) for s in key])
    #     return res % self.max_len

    def set(self, key, value):
        index = hash(key) % self.max_len
        self.table[index].append((key, value))

    def get(self, key):
        index = hash(key) % self.max_len
        value = self.table[index]

        if not value:
            return None

        for v in value:
            if v[0] == key:
                return v[1]

        return None


if __name__ == "__main__":
    capital = HashTable()
    country = ["Korea", "America", "China", "England", "turkiye"]
    city = ["Seoul", "Washington", "Beijing", "London", "Ankara"]

    for co, ci in zip(country, city):
        capital.set(co, ci)

    print("해시 테이블의 상태")
    print("=================")
    for i, v in enumerate(capital.table):
        print(i, v)
    print()

    print("======================")
    print(f"Capital of America = {capital.get('America')}")
    print(f"Capital of Korea = {capital.get('Korea')}")
    print(f"Capital of England = {capital.get('England')}")
    print(f"Capital of China = {capital.get('China')}")
    print(f"Capital of Japan = {capital.get('Japan')}")
    print(f"Capital of turkiye = {capital.get('turkiye')}")
```

```
# 키가 중복될 때 덮어쓰는 방법
# 실제로는 새로운 값이 들어오면 기존값을 덮어쓰는 방식이 일반적임.
def set(self, key, value):
    index = hash(key) % self.max_len
    for i, (k, v) in enumerate(self.table[index]):
        if k == key:  # 기존 키의 값을 갱신
            self.table[index][i] = (key, value)
            return
    self.table[index].append((key, value))  # 새 키 - 값 쌍 추가
```

```

# 문자열의 가장 먼저 나오는 유일한 문자 찾기(이중 반복문으로 풀기)
# 실무에서 많이 쓰진 않지만 , 차이를 알기 위해 구현
# 많이 안쓰는 이유 : 작은 코드에서 는 잘 체감 못하지만, 큰 프로젝트에서는 느려짐이 체감됨
def first_unique_char(s:str) -> int:
  n : int = len(s)
  check: list[int] = [0] * n  # 이미 확인한 문자 표시
  for i, ch in enumerate(s):
    if check[i]:  # 이미 중복으로 확인한 문자 스킵
       continue
    check[i] = 1
    is_unique = True
    for j in range(i+1, n ):  # 현재 문자 이후 탐색
      if ch == s[j]:
        check[j] = 1
        is_unique = False
        break             # 첫 중복 발생시 중단
    if is_unique:
        return i
  return -1


# 테스트 코드
for s in ("leetcode", "loveleetcode", "aabb"):
  print(first_unique_char(s))

```

```
# 딕셔너리 ver.
# 문자열에서 한 번만 등장하는 첫 문자의 인덱스 찾기

def first_unique_char_dic(s: str) -> int:
    # 1단계: 각 문자 등장 횟수 카운트
    count: dict[str, int] = {}
    for ch in s:
        if ch not in count:
            count[ch] = 1  # 처음 나온 문자면 1로 초기화
        else:
            count[ch] += 1  # 이미 존재하는 문자면 카운트 증가

    # 2단계: 검토 (완성된 카운트 보면서 확인)
    for i, c in enumerate(s):
        if count[c] == 1:  # 딱 한 번만 나온 문자면
            return i  # 그 문자의 인덱스 리턴

    # 3단계: 한 번만 나온 문자가 없으면 -1 리턴
    return -1

"""
첫 번째 for 문에서 문자의 등장횟수를 카운트.
두 번째 for 문에서 카운트가 1인 문자를 찾음.
"""
```

```

# 합이 0 인 부분 배열 중 가장 긴 것을 찾기
# ver. 딕셔너리

from collections import defaultdict

def maxlen_dic(arr: list[int]) -> int:
  total: int = 0
  dic: dict[int, list[int]] = defaultdict(list) # # {누적합: [인덱스들]}
  dic[0] = [-1]  # 시작점 0부터 시작하기 위해 사전에 -1로 초기화


  # 각 위치의 누적합 계산 및 저장
  for i in range(len(arr)):
    total += arr[i]   #  현재까지 누적합
    dic[total].append(i)  # 누적합이 나온 위치 기록
    # 같은 누적합이 나온 위치들 중 최대 거리 계산
  ans = 0
  for v in dic.values():
    ans = max(ans, v[-1] - v[0])  # (가장 먼 위치 - 가장 가까운 위치) # -1, 처음나온 위치 , 0이 마지막으로 나온 위치
  print(dict(dic))  # 주석해제
  return ans

#테스트 코드
A = [15, -2, 2, -8, 1, 7, 10,23]
print(maxlen_dic(A))

'''
빼는 개념이 아닌
가장 긴 것을 찾는다 --> 가장 멀리 떨어진 구간의 위치를 찾는거임
'''
```

```

# 합이 0인 부분 배열중 가장 긴것찾기 (딕셔너리 이용한 함수 개선)
# 첫 for 문에서 total 이 사전에 있으면 ans 를 바로 갱신
# for 문을 더 사용하지 않아도 되고 값을 리스트 로 만들지 않아도 됨


def maxlen_dic2(arr: list[int]) -> int:
    dic: dict[int, int] = {0: -1}
    ans: int = 0
    total: int = 0
    for i in range(len(arr)):
        total += arr[i]
        if total in dic:
            ans = max(ans, i - dic[total])
        else:
          dic[total] = i

    print(dict(dic))
    return ans


# 테스트코드
A = [15, -2, 2, -8, 1, 7, 10, 23]

print(maxlen_dic2(A))
```

```
"""
영어 끝말 잇기
1부터 n 까지 번호가 붙어있는 n 명의 사람이 영어 끝말잇기를 하고 있스빈다.
사람의 수 n 과 사람들이 순서대로 말한 단어 words 가 매개변수로 주어질 때 가장 먼저 탈락하는 사람의 번호와 그 사람이
자신의 몇 번째 차례에 탈락하는지를 구해서
return 하도록 solution 함수를 완성해 주세요 .

탈락사유 는 둘중 하나
1. 앞단어 의 끝 글자로 시작하는 단어가 아닐 때
2. 이미 나온 단어일 때

풀이:
첫째 탈락 사유 인지는 앞에 나왔던 단어의 끝 글자를 저장하고,
현재 단어의 첫글자와 비교.

둘째 탈락사유에 해당되는지는 집합을 이용하여 이민 나온 단어인지 확인
"""
# list를 이용한 풀이 , 개선 ver.


def solution(n: int, words: list[str]) -> list[int]:
    check: set[str] = set()  # 이미 나온 단어 저장
    tail: str = words[0][0]  # 첫 단어의 첫 글자 (시작점)
    for i, word in enumerate(words):
        # 1. 앞단어 의 끝 글자로 시작하는 단어가 아닐 때
        # 2. 이미 나온 단어일 때
        if word in check or tail != word[0]:
            return [i % n + 1, i // n + 1]  # 탈락
        check.add(word)  # 현재 단어를 기록
        tail = word[-1]  # 다음 사람이 시작해야 할 글자로 업데이트
    return [0, 0]


# 테스트 코드
n = 3
words = ["tank", "kick", "know", "wheel", "land", "dream", "mother", "robot", "tank"]
print(solution(n, words))
```

```
'''
배열로 이진트리 표현 - 자식노드 찾기

부모와 왼쪽 자식 오른쪽 자식을 출력
왼쪽자식와 오른쪽 자식노드의 인덱스를 계산 후
각 인덱스가 배열길이보다 작고 , 해당 배열의 값이 None이 아님 출력
배열 인덱스 규칙:
- 부모 인덱스: i
- 왼쪽 자식: 2*i + 1
- 오른쪽 자식: 2*i + 2
'''

tree = ["A", "B", "C", "D", "E", "F", None, "G"]

n = len(tree)
for i in range(n):
    if tree[i]:  # 노드가 존재하면
        print(f"Parent: {tree[i]}", end=", ")

        left = 2 * i + 1  # 왼쪽 자식 인덱스
        right = 2 * i + 2  # 오른쪽 자식 인덱스

        if left < n and tree[left] is not None:  # 왼쪽 자식 존재 체크
            print(f"Left: {tree[left]}", end=", ")

        if right < n and tree[right] is not None:  # 오른쪽 자식 존재 체크
            print(f"Right: {tree[right]}", end=" ")

        print()
```

```
'''
부모 노드 찾기

배열을 거꾸로 순회하며 각 노드의 부모를 출력
부모 인덱스 공식: (i - 1) // 2
'''

tree = ["A", "B", "C", "D", "E", "F", None, "G"]
n = len(tree)

for i in range(n - 1, 0, -1):  # 거꾸로 순회 (인덱스 1부터)
    if tree[i]:  # 노드가 존재하면
        parent_idx = (i - 1) // 2  # 부모 인덱스 계산
        print(f"Parent of {tree[i]} -> {tree[parent_idx]}")
```

```
'''
재귀 순회 함수 만들기

전위 순회 (Pre-order): 부모 → 왼쪽 → 오른쪽

재귀 종료 조건: 인덱스가 트리 크기를 넘어가면 종료
방문 처리: 현재 노드 출력
왼쪽/오른쪽 자식이 범위 안에 있고 None이 아니면 재귀 호출
'''

# 전위 순회
def preorder(tree, i=0):
    if i < len(tree):
        print(tree[i], end=" ")  # 현재 노드 방문

        left = 2 * i + 1  # 왼쪽 자식 인덱스
        right = left + 1  # 오른쪽 자식 인덱스

        if left < len(tree) and tree[left] is not None:  # 왼쪽 자식 존재 체크
            preorder(tree, left)

        if right < len(tree) and tree[right] is not None:  # 오른쪽 자식 존재 체크
            preorder(tree, right)


# 테스트
tree = ["A", "B", "C", "D", "E", "F", None, "G"]
preorder(tree)
```

```
'''
전위 순회 함수 (방문 결과 리스트 반환)

출력 대신 리스트로 반환하도록 변경
res 리스트에 현재 노드를 담고, 왼쪽/오른쪽 재귀 결과를 += 로 합침
'''

def preorder(tree, i=0):
    if i < len(tree):
        res = [tree[i]]  # 현재 노드를 리스트에 담음

        left = 2 * i + 1  # 왼쪽 자식 인덱스
        right = left + 1  # 오른쪽 자식 인덱스

        if left < len(tree) and tree[left] is not None:  # 왼쪽 자식 존재 체크
            res += preorder(tree, left)

        if right < len(tree) and tree[right] is not None:  # 오른쪽 자식 존재 체크
            res += preorder(tree, right)

        return res


# 테스트
tree = ["A", "B", "C", "D", "E", "F", None, "G"]
print(preorder(tree))
```

```
'''
중위 순회 함수 만들기

중위 순회 (In-order): 왼쪽 → 부모 → 오른쪽
내부 함수로 재귀 구현, 외부에서 res 리스트 관리
'''

def inorder(tree):
    def _inorder(i=0):  # 내부 재귀 함수
        if i >= len(tree) or tree[i] is None:  # 종료 조건
            return

        _inorder(2 * i + 1)  # 왼쪽 먼저
        res.append(tree[i])  # 현재 노드
        _inorder(2 * i + 2)  # 오른쪽 나중에

    res = []  # 결과 리스트
    _inorder()
    return res


# 테스트
tree = ["A", "B", "C", "D", "E", "F", None, "G"]
print(inorder(tree))
```

```
'''
후위 순회 함수

후위 순회 (Post-order): 왼쪽 → 오른쪽 → 부모
내부 함수로 재귀 구현, 외부에서 res 리스트 관리
'''

def postorder(tree):
    def _postorder(i=0):  # 내부 재귀 함수
        if i >= len(tree) or tree[i] is None:  # 종료 조건
            return

        _postorder(2 * i + 1)  # 왼쪽 먼저
        _postorder(2 * i + 2)  # 오른쪽 다음
        res.append(tree[i])  # 현재 노드 마지막

    res = []  # 결과 리스트
    _postorder()
    return res


# 테스트
tree = ["A", "B", "C", "D", "E", "F", None, "G"]
print(postorder(tree))
```
