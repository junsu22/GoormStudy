```
"""
해시테이블에 대한 예제를 구현하고 과정을 설명하세요
해시테이블의 충돌처리기법에 대해 설명하고 구현하세요
트리와 이진트리의 구성에 대해 이해하세요
트리 순회 방법에 따라 노드 추가/삭제 기법에 대해 예제를 구현하고 과정을 설명하세요

"""


# 해시테이블 클래스 만들기
# 게임으로 생각해보기
# <해시테이블이란 새로운 직업을 만들기>
class HashTable:
    def __init__(self, length=5):  # 캐릭터 생성할때 하는 기본세팅(자신, 인벤5칸)
        self.max_len = length  # 인벤토리 크기 정하기
        self.table = [
            [] for _ in range(self.max_len)
        ]  # 빈 칸들 만들기. 처음시작하면 아무것도 없으니까.

    # 설명 : 새로운 직업을 만들기 위한 기본셋팅, 빈인벤토리 5칸을 만들었다

    # 크기 줄여주기
    def _hash(self, key):  # "_"는 내부함수를 뜻함.
        res = sum([ord(s) for s in key])
        # 1. ord : 문자를 숫자로 바꿔주는 함수
        # 2. sum : 숫자를 전부 다 더한다.

        # 인벤토리 칸 개수로 나눈 나머지 = 저장할 칸 번호
        # 나누는 이유 : 숫자를 다 합산해서 엄청 큰 숫자가 나오지만
        # 만든 칸은 5개. 큰 숫자 (ex.100 번째 칸이 없어서.
        # 범위 안에 들어 갈 수 있게 나눠주는 거임 (0~4 안에 들어 갈수 있도록)
        # 다 더한값 % 만든 인벤 의 개수
        return res % self.max_len

    # 설명 : 템들이 모두 칸에 들어 갈 수 있도록 나눠서 크기를 줄여줬다.

    # 세팅
    def set(self, key, value):
        # 어느 칸에 넣을지 게산
        index = self._hash(key)
        # 그 칸에 템과 수량을 추가한다.
        self.table[index].append((key, value))

    # 템과 수량을 계산해서 몇번째 슬롯에 넣을지 정했다.

    # 지금 까지 한 일 .
    # 캐릭터 생성해서 인벤만들어주고 물약 계산해서 슬롯에 넣어줬음.

    def get(self, key):
        index = self._hash(key)  # 몇번째 칸에 있을지 계산
        value = self.table[index]  # 그 칸 전체를 가져온다

        if not value:  # 값이 비었는지
            return None  # 없다면 None 리턴

        for v in value:  # 그 칸의 아이템들 하나씩 확인
            if v[0] == key:  # 이름이 같은지 확인 (충돌 시 여러 개 중에서 찾기)
                return v[1]  # 수량 리턴

        return None  # 다 찾아봤지만 없음

    # 지금까지 한 일:
    # 캐릭터 만들고, 인벤 준비하고, 아이템 넣고(set), 아이템 찾기(get) 완성!
    # 충돌 처리: 같은 칸에 여러 개 들어가면 리스트로 저장하고, 찾을 때 하나씩 확인


# 앞에서 구현한 클래스 사용
if __name__ == "__main__":
    capital = HashTable()  # 객체 생성
    country = [
        "Korea",
        "America",
        "China",
        "England",
        "turkiye",
    ]  # 국가이름 5개 , 도시 이름 5개 매칭
    city = ["Seoul", "Washington", "Beijing", "London", "Ankara"]

    for co, ci in zip(country, city):
        capital.set(co, ci)
        # co , ci = ("korea" , "seoul")  해시 테이블에 저장
        #     ("merica", "Washington") ... 루프가 돌면서 해시 테이블 에 저장
    print("해시 테이블의 상태")
    print("=================")
    for i, v in enumerate(capital.table):
        # i : 인덱스(0 ,1, 2, 3, 4)
        # v : 값
        print(i, v)
    print()

    print("======================")
    # 문자를 숫자로 바꾼 전체 합을 다 더해서 5로 나누었을때 나머지는 0
    # 'A'+'m'+'e'...  65 + 109 + 101 .. %5 = 0 America = 0번째
    print(f"Capital of America = {capital.get('America')}")
    print(
        f"Capital of Korea = {capital.get('Korea')}"
    )  # 498% 5 = 3 나머지 3 (중국 과 충돌)
    print(f"Capital of England = {capital.get('England')}")
    print(f"Capital of China = {capital.get('China')}")  # 483 % 5 = 3 나머지 3
    # 위에 country , city 에서 japan은 선언 하지 않아 결과는 None
    print(f"Capital of Japan = {capital.get('Japan')}")
    print(f"Capital of turkiye = {capital.get('turkiye')}")
```

```
# 해시테이블 클래스 만들기 - 다른 방법
# 내장함수 hash 이용하기 (파이썬에 내장된 함수)

class HashTable:
    def __init__(self, length=5):
        self.max_len = length
        self.table = [[] for _ in range(self.max_len)]

    # 직접 만든 _hash 는 쓰지 않음. 내장 된 함수를 이용할 거임.
    # def _hash(self, key):
    #     res = sum([ord(s) for s in key])
    #     return res % self.max_len

    def set(self, key, value):
        index = hash(key) % self.max_len
        self.table[index].append((key, value))

    def get(self, key):
        index = hash(key) % self.max_len
        value = self.table[index]

        if not value:
            return None

        for v in value:
            if v[0] == key:
                return v[1]

        return None


if __name__ == "__main__":
    capital = HashTable()
    country = ["Korea", "America", "China", "England", "turkiye"]
    city = ["Seoul", "Washington", "Beijing", "London", "Ankara"]

    for co, ci in zip(country, city):
        capital.set(co, ci)

    print("해시 테이블의 상태")
    print("=================")
    for i, v in enumerate(capital.table):
        print(i, v)
    print()

    print("======================")
    print(f"Capital of America = {capital.get('America')}")
    print(f"Capital of Korea = {capital.get('Korea')}")
    print(f"Capital of England = {capital.get('England')}")
    print(f"Capital of China = {capital.get('China')}")
    print(f"Capital of Japan = {capital.get('Japan')}")
    print(f"Capital of turkiye = {capital.get('turkiye')}")
```
