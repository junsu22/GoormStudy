```python

# 사이킷런 이용해서 iris데이터 시각화하기
# DecisionTreeClassifier 학습모델
# 라이브러리 호출
from sklearn.datasets import load_iris
from sklearn.tree import DecisionTreeClassifier, export_graphviz
from graphviz import Source

# iris 데이터 로드
iris = load_iris(as_frame=True)

X = iris.data
y = iris.target

# 최대 깊이는 2  , 랜덤시드(난수), 새로 고침 할떄 마다 값이 변해서 고정
tree_clf = DecisionTreeClassifier(max_depth=2, random_state=42)
tree_clf.fit(X, y)  # X, y를 이용해 트리를 학습한다.
# graphviz 함수 이용
dot_data = export_graphviz(
    tree_clf,
    out_file=None,  # 파일 저장 안함 . 코랩에서 바로 시각화
    feature_names=iris.feature_names,  # 어떻게 분류했는지 표시
    class_names=iris.target_names,  # 꽃의 이름을 출력
    filled=True,  # 노드에 클래스 별 색 채우기
    rounded=True,  # 노드 박스를 둥글게
    special_characters=True,  # 특문 처리
)

graph = Source(dot_data)
graph  # 코랩 마지막 셀 에서 그림이 자동 랜더링

# PNG , PDF 저장 가능
graph.format = "png"
graph.render("iris", cleanup=True)
```
